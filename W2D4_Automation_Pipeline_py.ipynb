```python
import time
import sqlite3
import requests
import json

# إعدادات الـ API (استخدمنا API عام ومجاني كمثال لجلب بيانات)
API_URL = "https://jsonplaceholder.typicode.com/posts/1"

def run_pipeline(cycle_number):
    try:
        print(f"\n--- بدء تنفيذ الدورة رقم {cycle_number} ---")

        # 1. الخطوة الأولى: جلب البيانات من الـ API
        response = requests.get(API_URL)

        if response.status_code == 200:
            data = response.json()
            print("1. تم جلب البيانات من الـ API بنجاح (Status 200).")
        else:
            print(f"فشل جلب البيانات، كود الخطأ: {response.status_code}")
            return

        # 2. الخطوة الثانية: تخزين البيانات في قاعدة البيانات store.db
        conn = sqlite3.connect('store.db')
        cursor = conn.cursor()

        # التأكد من إنشاء الجدول إن لم يكن موجوداً
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS pipeline_logs (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                api_data TEXT NOT NULL,
                timestamp DATETIME DEFAULT CURRENT_TIMESTAMP
            )
        ''')

        # إدخال البيانات
        cursor.execute("INSERT INTO pipeline_logs (api_data) VALUES (?)", (json.dumps(data),))
        conn.commit()
        conn.close()
        print("2. تم تخزين البيانات في قاعدة البيانات (store.db) بنجاح.")

        print(f"تم تنفيذ الدورة رقم {cycle_number} بنجاح ✅")

    except Exception as e:
        print(f"حدث خطأ أثناء تنفيذ الـ Pipeline: {e}")

def main():
    print("--- بدء تشغيل نظام الـ Automation Pipeline ---")
    cycle = 1

    # 4. الأتمتة: تكرار تشغيل الدالة تلقائياً كل 30 ثانية
    try:
        # لتجربة دورة واحدة أو دورتين على كولاب بسرعة، أو تركها تعمل
        for cycle in range(1, 4):  # سينفذ 3 دورات كمثال تجريبي ويتوقف بلطف
            run_pipeline(cycle)
            if cycle < 3:
                print("انتظار 30 ثانية للدورة القادمة...")
                time.sleep(30)
    except KeyboardInterrupt:
        print("\nتم إيقاف الـ Pipeline بواسطة المستخدم.")

if __name__ == "__main__":
    main()
```
